Problem 1 — NumPy Warm-up
Open a new notebook (Colab or VS Code). Without copy-pasting the examples above:

1. Create an array of numbers 1–20. Example: [1, 2, 3] using numpy 
2. Select only even numbers using both slicing and filtering. Example: case one using slicing based on index [1, 3, 5, ...]. case two filtering based on value. 
3. Compute the mean and standard deviation Example: mean = 2, standard deviation = 1
4. Create a 4×4 matrix of zeros, then set the diagonal values to 1. 

```python
import numpy as np

# 1. Create an array of numbers 1–10.
arr = np.arange(1, 11)
print("Array 1-10:", arr)

# 2. Select only even numbers using slicing and filtering.
# Using slicing (even numbers are at odd indices in 0-based indexing: 1,3,5,...)
even_slicing = arr[1::2]
print("Even numbers using slicing:", even_slicing)

# Using filtering
even_filtering = arr[arr % 2 == 0]
print("Even numbers using filtering:", even_filtering)

# 3. Compute the mean and standard deviation.
mean_val = arr.mean()
std_val = arr.std()
print("Mean:", mean_val)
print("Standard deviation:", std_val)

# 4. Create a 4×4 matrix of zeros, then set the diagonal values to 1.
identity_like = np.zeros((4, 4), dtype=int)
np.fill_diagonal(identity_like, 1)
print("4x4 identity-like matrix:")
print(identity_like)
```
