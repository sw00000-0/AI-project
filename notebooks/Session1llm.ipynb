Problem 1 — NumPy Warm-up
Open a new notebook (Colab or VS Code). Without copy-pasting the examples above:

Create an array of numbers 1–20
Select only even numbers using slicing or filtering
Compute the mean and standard deviation
Create a 4×4 matrix of zeros, then set the diagonal values to 1

```python
import numpy as np

arr = np.arange(1, 21)
arr
```

```python
even_values = arr[arr % 2 == 0]
even_values
```

```python
evens = arr[::2][1:] if arr[0] % 2 != 0 else arr[::2]
evens
```

```python
even_index_elements = np.where(arr % 2 == 0, arr, 0)
even_index_elements = even_index_elements[even_index_elements != 0]
even_index_elements
```

```python
mean_val = np.mean(arr)
std_val = np.std(arr)
mean_val, std_val
```

```python
matrix = np.zeros((4, 4))
np.fill_diagonal(matrix, 1)
matrix
```
